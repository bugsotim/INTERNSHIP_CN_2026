```python
import pandas as pd

# รวม DataFrames df_2024, df_2025, และ df_2026 เข้าด้วยกัน
# ตรวจสอบว่า DataFrames มีอยู่ก่อนที่จะทำการรวม

all_dfs_to_concat = []

if 'df_2024' in locals() and not df_2024.empty:
    all_dfs_to_concat.append(df_2024)
    print(f"df_2024 มี {df_2024.shape[0]} แถว")
else:
    print("ไม่พบ df_2024 หรือเป็น DataFrame ว่าง")

if 'df_2025' in locals() and not df_2025.empty:
    all_dfs_to_concat.append(df_2025)
    print(f"df_2025 มี {df_2025.shape[0]} แถว")
else:
    print("ไม่พบ df_2025 หรือเป็น DataFrame ว่าง")

if 'df_2026' in locals() and not df_2026.empty:
    all_dfs_to_concat.append(df_2026)
    print(f"df_2026 มี {df_2026.shape[0]} แถว")
else:
    print("ไม่พบ df_2026 หรือเป็น DataFrame ว่าง")

if all_dfs_to_concat:
    shopee = pd.concat(all_dfs_to_concat, ignore_index=True)
    print(f"\nDataFrame 'shopee' สร้างแล้วด้วย {shopee.shape[0]} แถว และ {shopee.shape[1]} คอลัมน์")
else:
    print("ไม่พบ DataFrame ที่จะนำมารวมกัน ไม่มี DataFrame 'shopee' ถูกสร้างขึ้น")
```
