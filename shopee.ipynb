```python
import re

# Define the keywords to search for
keywords_to_search = [
    '[ของแถม]',
    '[สินค้าแถม ห้ามกดสั่งซื้อ]'
]

# Escape special characters in each keyword for regex and join them with '|'
escaped_keywords_for_filter = [re.escape(k) for k in keywords_to_search]
pattern_for_filter = '|'.join(escaped_keywords_for_filter)

# Filter the DataFrame
filtered_items = shopee1[shopee1['ชื่อสินค้า'].str.contains(pattern_for_filter, na=False, regex=True)]

print(f"จำนวนรายการที่พบ: {len(filtered_items)}")
# Display the filtered DataFrame
filtered_items
```

```python
import re

# Define the keywords to remove
keywords_to_remove = [
    '[ของแถม]',
    '[สินค้าแถม ห้ามกดสั่งซื้อ]'
]

# Escape special characters in each keyword for regex and join them with '|'
escaped_keywords_for_removal = [re.escape(k) for k in keywords_to_remove]
pattern_for_removal = '|'.join(escaped_keywords_for_removal);

# Create a boolean mask for rows that contain any of the keywords
# The '~' operator negates the mask, selecting rows that do NOT contain the keywords
shopee1 = shopee1[~shopee1['ชื่อสินค้า'].str.contains(pattern_for_removal, na=False, regex=True)];

print(f"DataFrame 'shopee1' หลังจากลบแถว: {shopee1.shape[0]} แถว และ {shopee1.shape[1]} คอลัมน์")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = shopee2.copy()

# =========================
# DATE COLUMN
# =========================
date_cols = [
    'วันที่ทำการสั่งซื้อ',
    'เวลาการชำระสินค้า',
    'วันที่คาดว่าจะทำการจัดส่งสินค้า',
    'เวลาส่งสินค้า',
    'เวลาที่ทำการสั่งซื้อสำเร็จ'
]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# =========================
# NUMERIC COLUMN
# =========================
num_cols = [
    'ราคาตั้งต้น',
    'ราคาขาย',
    'จำนวน',
    'จำนวนที่ส่งคืน',
    'ยอดชำระเงิน',
    'ส่วนลดจาก_Shopee',
    'โค้ดส่วนลดชำระโดยผู้ขาย',
    'โค้ดส่วนลดชำระโดย_Shopee_(เช่น_โค้ดจากโปรแกรม_ร้านโค้ดคุ้ม,_โค้ดส่วนลด_Shopee,_โค้ดส่วนลด_Shopee_Mall)',
    'ส่วนลดจากการใช้เหรียญ',
    'ค่าคอมมิชชั่น',
    'ราคาสินค้าที่ชำระโดยผู้ซื้อ_(THB)',
    'ค่าบริการ',
    'จำนวนเงินทั้งหมด',
    'ค่าจัดส่งโดยประมาณ',
    'ค่าธรรมเนียม_(%)'
]

for col in num_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(',', '')
        .str.replace('฿', '')
    )

    df[col] = pd.to_numeric(df[col], errors='coerce')

# =========================
# CREATE NEW COLUMN
# =========================

df['month'] = df['วันที่ทำการสั่งซื้อ'].dt.month
df['weekday'] = df['วันที่ทำการสั่งซื้อ'].dt.day_name()
df['hour'] = df['เวลาการชำระสินค้า'].dt.hour

df['ส่วนลดรวม'] = (
    + df['โค้ดส่วนลดชำระโดยผู้ขาย'].fillna(0)
    + df['โค้ดส่วนลดชำระโดย_Shopee_(เช่น_โค้ดจากโปรแกรม_ร้านโค้ดคุ้ม,_โค้ดส่วนลด_Shopee,_โค้ดส่วนลด_Shopee_Mall)'].fillna(0)
)

df['gross_sales'] = df['ราคาขาย'] * df['จำนวน']

df['net_sales'] = (
    df['จำนวนเงินทั้งหมด']
    - df['ค่าคอมมิชชั่น']
    - df['ค่าบริการ']
)

df['estimated_profit'] = (
    df['จำนวนเงินทั้งหมด']
    - df['ส่วนลดรวม']
    - df['ค่าคอมมิชชั่น']
    - df['ค่าบริการ']
    - df['ค่าจัดส่งโดยประมาณ']
)
```

```python

# Return Flag
df['return_flag'] = np.where(df['จำนวนที่ส่งคืน'] > 0, 1, 0)

# Shipping Delay
df['processing_day'] = (
    df['เวลาส่งสินค้า']
    - df['เวลาการชำระสินค้า']
).dt.days
```

```python
# =========================================
# 3. MONTHLY SALES TREND
# =========================================

monthly_sales = (
    df.groupby(['year', 'month'])['net_sales']
    .sum()
    .reset_index()
)

monthly_sales['year_month'] = (
    monthly_sales['year'].astype(str)
    + '-'
    + monthly_sales['month'].astype(str)
)

plt.figure(figsize=(14,5))

plt.plot(
    monthly_sales['year_month'],
    monthly_sales['net_sales']
)

plt.xticks(rotation=90)

plt.title('Monthly Sales Trend')
plt.ylabel('Net Sales')

plt.show()
```
