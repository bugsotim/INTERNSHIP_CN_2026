```python
import re

# Define the keywords to search for
keywords_to_search = [
    '[ของแถม]',
    '[สินค้าแถม ห้ามกดสั่งซื้อ]'
]

# Escape special characters in each keyword for regex and join them with '|'
escaped_keywords_for_filter = [re.escape(k) for k in keywords_to_search]
pattern_for_filter = '|'.join(escaped_keywords_for_filter)

# Filter the DataFrame
filtered_items = shopee1[shopee1['ชื่อสินค้า'].str.contains(pattern_for_filter, na=False, regex=True)]

print(f"จำนวนรายการที่พบ: {len(filtered_items)}")
# Display the filtered DataFrame
filtered_items
```

```python
import re

# Define the keywords to remove
keywords_to_remove = [
    '[ของแถม]',
    '[สินค้าแถม ห้ามกดสั่งซื้อ]'
]

# Escape special characters in each keyword for regex and join them with '|'
escaped_keywords_for_removal = [re.escape(k) for k in keywords_to_remove]
pattern_for_removal = '|'.join(escaped_keywords_for_removal);

# Create a boolean mask for rows that contain any of the keywords
# The '~' operator negates the mask, selecting rows that do NOT contain the keywords
shopee1 = shopee1[~shopee1['ชื่อสินค้า'].str.contains(pattern_for_removal, na=False, regex=True)];

print(f"DataFrame 'shopee1' หลังจากลบแถว: {shopee1.shape[0]} แถว และ {shopee1.shape[1]} คอลัมน์")
```

```python
# Define the columns used in the calculation
selling_price_col = 'ราคาขาย'
seller_discount_col = 'โค้ดส่วนลดชำระโดยผู้ขาย'
shopee_discount_col = 'โค้ดส่วนลดชำระโดย Shopee (เช่น โค้ดจากโปรแกรม ร้านโค้ดคุ้ม, โค้ดส่วนลด Shopee, โค้ดส่วนลด Shopee Mall)'

# Ensure the columns are numeric, coercing errors will turn non-numeric values into NaN
shopee1[selling_price_col] = pd.to_numeric(shopee1[selling_price_col], errors='coerce').fillna(0)
shopee1[seller_discount_col] = pd.to_numeric(shopee1[seller_discount_col], errors='coerce').fillna(0)
shopee1[shopee_discount_col] = pd.to_numeric(shopee1[shopee_discount_col], errors='coerce').fillna(0)

# Calculate the total discounts
total_discounts = shopee1[seller_discount_col] + shopee1[shopee_discount_col]

# Calculate Discount1, handling division by zero
# Where 'ราคาขาย' is 0, Discount1 will be 0 to avoid errors
shopee1['Discount1'] = (
    shopee1.apply(lambda row:
        ((row[selling_price_col] - (row[seller_discount_col] + row[shopee_discount_col])) / row[selling_price_col]) * 100
        if row[selling_price_col] != 0 else 0,
        axis=1
    )
)

print(f"Column 'Discount1' created. Here's a sample of the new column:")
print(shopee1[[selling_price_col, seller_discount_col, shopee_discount_col, 'Discount1']].head())
```
