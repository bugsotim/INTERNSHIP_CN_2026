```python
df_01_2026_cleaned = df_01_2026.iloc[1:].copy()
df_02_2026_cleaned = df_02_2026.iloc[1:].copy()
df_03_2026_cleaned = df_03_2026.iloc[1:].copy()

df2026 = pd.concat([df_01_2026_cleaned, df_02_2026_cleaned, df_03_2026_cleaned], ignore_index=True)
```

Now, I'll convert the `Order Creation Time` column to a datetime object and then sort the `df2026` DataFrame.

```python
df2026['Created Time'] = pd.to_datetime(df2026['Created Time'], format='%d/%m/%Y %H:%M:%S')
df2026 = df2026.sort_values(by='Created Time').reset_index(drop=True)

display(df2026.head())
```

```python
df_2024_Q1_cleaned = df_2024_Q1.iloc[1:].copy()
df_2024_Q234_cleaned = df_2024_Q234.iloc[1:].copy()

df2024 = pd.concat([df_2024_Q1_cleaned, df_2024_Q234_cleaned], ignore_index=True)
display(df2024.head())
```

# Calculate pay lag((timepayment − timecreate) ทุก order)

```python
ttdf['Created Time'] = pd.to_datetime(ttdf['Created Time'], errors='coerce') #การแปลงข้อมูลให้เป็น "วันที่"
ttdf['Paid Time'] = pd.to_datetime(ttdf['Paid Time'], errors='coerce') #ทำกับคอลัมน์ Paid Time
ttdf['Pay Lag'] = ttdf['Paid Time'] - ttdf['Created Time'] #คำนวณส่วนต่างเวลา (Pay Lag) เวลาจ่ายเงิน ลบด้วย เวลาที่สร้าง
display(ttdf[['Created Time', 'Paid Time', 'Pay Lag']]) #ดูเฉพาะ 3 คอลัมน์ที่เราสนใจ (Created Time, Paid Time, Pay Lag)
```
