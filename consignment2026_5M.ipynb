```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# --- Preprocessing (Ensure column names and types are correct) ---
# Clean column names: remove non-printable characters, strip whitespace, and replace multiple spaces
kis.columns = kis.columns.map(lambda x: ''.join(filter(str.isprintable, x)).strip())
kis.columns = kis.columns.str.replace(r'\s+', ' ', regex=True) # Standardize spaces

# Initialize quantity_col_name_found to None. It will hold the actual column name if found.
quantity_col_name_found = None

# First, check for an exact match to 'Sales Quantity' (after cleaning and stripping)
if 'Sales Quantity' in kis.columns:
    quantity_col_name_found = 'Sales Quantity'
else:
    # If not found directly, iterate and search for a case-insensitive match 'sales quantity'
    for col in kis.columns:
        if col.lower() == 'sales quantity':
            quantity_col_name_found = col
            break

    if quantity_col_name_found: # If an exact case-insensitive match was found but not in exact 'Sales Quantity' form
        if quantity_col_name_found != 'Sales Quantity':
            kis.rename(columns={quantity_col_name_found: 'Sales Quantity'}, inplace=True)
            quantity_col_name_found = 'Sales Quantity' # Update the variable to the new name
    else:
        # If no exact match (case-sensitive or insensitive) found, try broader search for 'quantity' as substring
        print(f"Warning: Exact column 'Sales Quantity' not found. Searching for 'quantity' substring.")
        for col in kis.columns:
            if 'quantity' in col.lower():
                quantity_col_name_found = col
                break

        if quantity_col_name_found:
            if quantity_col_name_found != 'Sales Quantity': # Rename only if it's not already the target name
                print(f"Found '{quantity_col_name_found}' as a potential quantity column. Renaming to 'Sales Quantity'.")
                kis.rename(columns={quantity_col_name_found: 'Sales Quantity'}, inplace=True)
                quantity_col_name_found = 'Sales Quantity' # Update the variable to the new name
        else:
            print(f"Error: No column related to 'Quantity' found in the DataFrame after extensive cleaning. Available columns: {kis.columns.tolist()}")
            # quantity_col_name_found remains None, which will be handled by subsequent if checks.

# Now, process the 'Sales Quantity' column if it was successfully identified
if quantity_col_name_found:
    kis['Sales Quantity'] = pd.to_numeric(kis['Sales Quantity'], errors='coerce')
    kis['Sales Quantity'] = kis['Sales Quantity'].fillna(0)
else:
    # If 'Sales Quantity' column could not be found, create a dummy column to avoid KeyErrors
    # This ensures that KPI calculations and plots don't fail, but show 0 or skip.
    kis['Sales Quantity'] = 0 # Dummy column, will cause total_quantity to be 0
    print("A dummy 'Sales Quantity' column with all zeros has been created as the original could not be identified.")

# Ensure 'Sales Date' is in datetime format and sort the DataFrame
kis['Sales Date'] = pd.to_datetime(kis['Sales Date'])
kis_sorted = kis.sort_values(by='Sales Date')

# Ensure the primary sales column is numeric
kis['ยอดหลังหัก Discount,VAT,GP(35%)'] = pd.to_numeric(kis['ยอดหลังหัก Discount,VAT,GP(35%)'], errors='coerce').fillna(0)

# --- 1. KPI Cards ---
print("\n--- KPI Cards ---")
total_sales = kis['ยอดหลังหัก Discount,VAT,GP(35%)'].sum()
total_quantity = kis['Sales Quantity'].sum()
avg_sales_per_item = total_sales / total_quantity if total_quantity > 0 else 0

print(f"Total Sales (after Discount, VAT, GP): {total_sales:,.2f} THB")
print(f"Total Quantity Sold: {total_quantity:,.0f} units")
print(f"Average Sales per Item: {avg_sales_per_item:,.2f} THB/unit")

# --- 2. Sales Trend ---
print("\n--- Sales Trend Plots ---")
plt.figure(figsize=(18, 7))

plt.subplot(1, 2, 1)
sns.lineplot(x='Sales Date', y='ยอดหลังหัก Discount,VAT,GP(35%)', data=kis_sorted)
plt.title('Sales Trend for KIS (After All Deductions)')
plt.xlabel('Sales Date')
plt.ylabel('Sales (THB)')
plt.grid(True)

# Only plot quantity if the column was successfully identified (quantity_col_name_found is True/not None)
if quantity_col_name_found:
    plt.subplot(1, 2, 2)
    sns.lineplot(x='Sales Date', y='Sales Quantity', data=kis_sorted)
    plt.title('Quantity Trend for KIS')
    plt.xlabel('Sales Date')
    plt.ylabel('Quantity Sold')
    plt.grid(True)
else:
    print("Skipping Quantity Trend plot as 'Sales Quantity' column was not identified.")

plt.tight_layout()
plt.show()

# --- 3. Top Brand ---
print("\n--- Top Brands by Sales and Quantity ---")
top_brands_sales = kis.groupby('Brand Name')['ยอดหลังหัก Discount,VAT,GP(35%)'].sum().nlargest(5)
top_brands_quantity = kis.groupby('Brand Name')['Sales Quantity'].sum().nlargest(5) if quantity_col_name_found else pd.Series([])

plt.figure(figsize=(18, 6))

plt.subplot(1, 2, 1)
sns.barplot(x=top_brands_sales.values, y=top_brands_sales.index, palette='viridis')
plt.title('Top 5 Brands by Sales')
plt.xlabel('Total Sales (THB)')
plt.ylabel('Brand Name')
plt.grid(axis='x')

if quantity_col_name_found:
    plt.subplot(1, 2, 2)
    sns.barplot(x=top_brands_quantity.values, y=top_brands_quantity.index, palette='magma')
    plt.title('Top 5 Brands by Quantity')
    plt.xlabel('Total Quantity Sold')
    plt.ylabel('Brand Name')
    plt.grid(axis='x')
else:
    print("Skipping Top Brands by Quantity plot as 'Sales Quantity' column was not identified.")

plt.tight_layout()
plt.show()

# --- 4. Top Store ---
print("\n--- Top Stores by Sales and Quantity ---")
top_stores_sales = kis.groupby('Store Name')['ยอดหลังหัก Discount,VAT,GP(35%)'].sum().nlargest(5)
top_stores_quantity = kis.groupby('Store Name')['Sales Quantity'].sum().nlargest(5) if quantity_col_name_found else pd.Series([])

plt.figure(figsize=(18, 6))

plt.subplot(1, 2, 1)
sns.barplot(x=top_stores_sales.values, y=top_stores_sales.index, palette='viridis')
plt.title('Top 5 Stores by Sales')
plt.xlabel('Total Sales (THB)')
plt.ylabel('Store Name')
plt.grid(axis='x')

if quantity_col_name_found:
    plt.subplot(1, 2, 2)
    sns.barplot(x=top_stores_quantity.values, y=top_stores_quantity.index, palette='magma')
    plt.title('Top 5 Stores by Quantity')
    plt.xlabel('Total Quantity Sold')
    plt.ylabel('Store Name')
    plt.grid(axis='x')
else:
    print("Skipping Top Stores by Quantity plot as 'Sales Quantity' column was not identified.")

plt.tight_layout()
plt.show()

# --- 5. Top SKU ---
print("\n--- Top SKUs by Sales and Quantity ---")
top_skus_sales = kis.groupby('SKU Name')['ยอดหลังหัก Discount,VAT,GP(35%)'].sum().nlargest(5)
top_skus_quantity = kis.groupby('SKU Name')['Sales Quantity'].sum().nlargest(5) if quantity_col_name_found else pd.Series([])

plt.figure(figsize=(18, 6))

plt.subplot(1, 2, 1)
sns.barplot(x=top_skus_sales.values, y=top_skus_sales.index, palette='viridis')
plt.title('Top 5 SKUs by Sales')
plt.xlabel('Total Sales (THB)')
plt.ylabel('SKU Name')
plt.grid(axis='x')

if quantity_col_name_found:
    plt.subplot(1, 2, 2)
    sns.barplot(x=top_skus_quantity.values, y=top_skus_quantity.index, palette='magma')
    plt.title('Top 5 SKUs by Quantity')
    plt.xlabel('Total Quantity Sold')
    plt.ylabel('SKU Name')
    plt.grid(axis='x')
else:
    print("Skipping Top SKUs by Quantity plot as 'Sales Quantity' column was not identified.")

plt.tight_layout()
plt.show()

# --- 6. Waterfall Discount -> GP (simplified breakdown) ---
print("\n--- Sales Breakdown (Gross to Net) ---")
# Ensure these columns are numeric as well
kis['Gross Sales'] = pd.to_numeric(kis['Gross Sales'], errors='coerce').fillna(0)
kis['Total Discount'] = pd.to_numeric(kis['Total Discount'], errors='coerce').fillna(0)
kis['ยอดหลังหัก Discount'] = pd.to_numeric(kis['ยอดหลังหัก Discount'], errors='coerce').fillna(0)
kis['ยอดหลังหักDiscount และVAT'] = pd.to_numeric(kis['ยอดหลังหักDiscount และVAT'], errors='coerce').fillna(0)
kis['ยอดหลังหัก Discount,VAT,GP(35%)'] = pd.to_numeric(kis['ยอดหลังหัก Discount,VAT,GP(35%)'], errors='coerce').fillna(0)

# Calculate aggregated values for the breakdown
total_gross_sales = kis['Gross Sales'].sum()
total_discount = kis['Total Discount'].sum() # This is typically a positive number representing deduction
total_after_discount = kis['ยอดหลังหัก Discount'].sum()
total_vat = total_after_discount - kis['ยอดหลังหักDiscount และVAT'].sum()
total_after_vat = kis['ยอดหลังหักDiscount และVAT'].sum()
total_gp_deduction = total_after_vat - kis['ยอดหลังหัก Discount,VAT,GP(35%)'].sum()
final_sales_value = kis['ยอดหลังหัก Discount,VAT,GP(35%)'].sum()

# Prepare data for plotting
breakdown_data = {
    'Metric': ['Gross Sales', 'Discount', 'After Discount', 'VAT', 'After VAT', 'GP Deduction', 'Final Sales (Net)'],
    'Value': [
        total_gross_sales,
        -total_discount, # Represent discount as a negative change
        total_after_discount,
        -total_vat,      # Represent VAT as a negative change
        total_after_vat,
        -total_gp_deduction, # Represent GP as a negative change
        final_sales_value
    ]
}
breakdown_df = pd.DataFrame(breakdown_data)

# Let's plot the cumulative value at each stage to show the 'waterfall'
waterfall_values = [
    total_gross_sales,
    total_after_discount,
    total_after_vat,
    final_sales_value
]
waterfall_labels = [
    'Gross Sales',
    'Sales After Discount',
    'Sales After Discount & VAT',
    'Final Sales (After GP)'
]

plt.figure(figsize=(10, 6))
sns.barplot(x=waterfall_values, y=waterfall_labels, palette='coolwarm')
plt.title('Sales Value Progression: Gross to Net')
plt.xlabel('Amount (THB)')
plt.ylabel('Stage')
plt.grid(axis='x')
plt.tight_layout()
plt.show()

# --- 7. Dept/SubDept Analysis ---
print("\n--- Department and Sub-Department Analysis ---")
dept_sales = kis.groupby('Dept')['ยอดหลังหัก Discount,VAT,GP(35%)'].sum().sort_values(ascending=False)
dept_quantity = kis.groupby('Dept')['Sales Quantity'].sum().sort_values(ascending=False) if quantity_col_name_found else pd.Series([])

sub_dept_sales = kis.groupby('Sub Dept')['ยอดหลังหัก Discount,VAT,GP(35%)'].sum().nlargest(10) # Top 10 for readability
sub_dept_quantity = kis.groupby('Sub Dept')['Sales Quantity'].sum().nlargest(10) if quantity_col_name_found else pd.Series([])

# Plot by Department
plt.figure(figsize=(18, 6))

plt.subplot(1, 2, 1)
sns.barplot(x=dept_sales.values, y=dept_sales.index, palette='viridis')
plt.title('Sales by Department')
plt.xlabel('Total Sales (THB)')
plt.ylabel('Department')
plt.grid(axis='x')

if quantity_col_name_found:
    plt.subplot(1, 2, 2)
    sns.barplot(x=dept_quantity.values, y=dept_quantity.index, palette='magma')
    plt.title('Quantity Sold by Department')
    plt.xlabel('Total Quantity Sold')
    plt.ylabel('Department')
    plt.grid(axis='x')
else:
    print("Skipping Quantity Sold by Department plot as 'Sales Quantity' column was not identified.")

plt.tight_layout()
plt.show()

# Plot by Sub-Department (Top N)
plt.figure(figsize=(18, 6))

plt.subplot(1, 2, 1)
sns.barplot(x=sub_dept_sales.values, y=sub_dept_sales.index, palette='viridis')
plt.title('Top 10 Sub-Departments by Sales')
plt.xlabel('Total Sales (THB)')
plt.ylabel('Sub-Department')
plt.grid(axis='x')

if quantity_col_name_found:
    plt.subplot(1, 2, 2)
    sns.barplot(x=sub_dept_quantity.values, y=sub_dept_quantity.index, palette='magma')
    plt.title('Top 10 Sub-Departments by Quantity Sold')
    plt.xlabel('Total Quantity Sold')
    plt.ylabel('Sub-Department')
    plt.grid(axis='x')
else:
    print("Skipping Top 10 Sub-Departments by Quantity Sold plot as 'Sales Quantity' column was not identified.")

plt.tight_layout()
plt.show()
```
