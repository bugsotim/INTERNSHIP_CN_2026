# โหลดไฟล์ข้อมูล

```python
df = pd.read_csv('/content/drive/MyDrive/ฝึกงาน CN/Sales_report_consignment 20.04.2026 to Dashboard.csv')
display(df.head())
```

## ลบแถวที่เป็นข้อมูลของปี 2025 ออก

```python
print(f"Number of rows before filtering: {len(df)}")

# Filter out rows where 'ปี-เดือน' column contains '2025'
df2 = df[~df['ปี-เดือน'].astype(str).str.contains('2025', na=False)]

print(f"Number of rows after filtering: {len(df)}")
display(df.head())
```

```python
df2
```

## ลบแถวที่คอลัมน์ สาขา มีค่าเป็น Warehouse BMA

```python
print(f"Number of rows in df2 before filtering: {len(df2)}")

df3 = df2[df2['สาขา'] != 'Warehouse BMA']

print(f"Number of rows in df2 after filtering: {len(df2)}")
display(df3.head())
```
